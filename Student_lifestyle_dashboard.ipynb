## 3. Feature Engineering

```python
# Efficiency
df['Study_Efficiency'] = df['Previous_GPA'] / df['Hours_Studied']

```

```python
df['Engagement_Score'] = df['Attendance'] + df['Hours_Studied']
```

```python
df['Attendance_norm'] = (df['Attendance'] - df['Attendance'].min()) / (df['Attendance'].max() - df['Attendance'].min())
```

```python
df['Hours_norm'] = (df['Hours_Studied'] - df['Hours_Studied'].min()) / (df['Hours_Studied'].max() - df['Hours_Studied'].min())
```

```python
df['Sleep_norm'] = (df['Sleep_Hours'] - df['Sleep_Hours'].min()) / (df['Sleep_Hours'].max() - df['Sleep_Hours'].min())
```

```python
df['Screen_norm'] = (df['Screen_Time'] - df['Screen_Time'].min()) / (df['Screen_Time'].max() - df['Screen_Time'].min())
```

```python
df['Stress_norm'] = (df['Stress_Level'] - df['Stress_Level'].min()) / (df['Stress_Level'].max() - df['Stress_Level'].min())
```

```python
df['Anxiety_norm'] = (df['Exam_Anxiety_Score'] - df['Exam_Anxiety_Score'].min()) / (df['Exam_Anxiety_Score'].max() - df['Exam_Anxiety_Score'].min())
```

Engagement Score was created by combining normalized attendance and study hours to represent overall academic involvement.

Lifestyle Score was derived using sleep hours and screen time to capture balance in daily habits.

Mental Stress Score combines stress level and exam anxiety to measure psychological pressure.

```python
df['Engagement_Score'] = df['Attendance_norm'] + df['Hours_norm']
```

```python
df['Lifestyle_Score'] = df['Sleep_norm'] - df['Screen_norm']
```

```python
df['Mental_Stress_Score'] = df['Stress_norm'] + df['Anxiety_norm']
```

```python
df[['Engagement_Score','Lifestyle_Score','Mental_Stress_Score']].describe()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

## 4. Exploratory Data Analysis

###1.Correlation Heatmap

The heatmap shows relationships between variables. Strong correlations help identify key factors influencing performance.

```python
plt.figure(figsize=(10,6))
sns.heatmap(df.corr(numeric_only=True), annot=True)
plt.title("Correlation Matrix")
plt.show()
```

###2: Engagement vs Grade

This plot shows how student engagement impacts grades.

```python
plt.figure()
sns.boxplot(x='Grade', y='Engagement_Score', data=df)
plt.title("Engagement vs Grade")
plt.show()
```

###3: Lifestyle vs Grade

This visualization highlights the effect of lifestyle balance on academic performance.

```python
plt.figure()
sns.boxplot(x='Grade', y='Lifestyle_Score', data=df)
plt.title("Lifestyle Impact on Grades")
plt.show()
```

###4: Stress vs Grade

This graph analyzes the impact of stress and anxiety on grades.

```python
plt.figure()
sns.boxplot(x='Grade', y='Mental_Stress_Score', data=df)
plt.title("Stress Impact on Grades")
plt.show()
```

```python
plt.figure()
sns.scatterplot(x='Hours_Studied', y='Previous_GPA', data=df)
plt.title("Study Hours vs GPA")
plt.show()
```

###5: Attendance vs Grade

Attendance is compared across grade categories to observe its influence.

```python
plt.figure()
sns.boxplot(x='Grade', y='Attendance', data=df)
plt.title("Attendance vs Grade")
plt.show()
```

```python
grade_counts = df['Grade'].value_counts()

plt.figure()
grade_counts.plot(kind='bar')
plt.title("Distribution of Student Grades")
plt.xlabel("Grade")
plt.ylabel("Number of Students")
plt.show()
```

###Insight:
The distribution shows how students are spread across grade categories. A higher concentration in top grades indicates strong overall performance, while a spread toward lower grades may highlight areas needing improvement.

```python
pd.crosstab(df['Part_Time_Job'], df['Grade']).plot(kind='bar')

plt.title("Impact of Part-Time Job on Grades")
plt.xlabel("Part-Time Job")
plt.ylabel("Number of Students")
plt.show()
```

###Insight:
Students with part-time jobs show differences in grade distribution, suggesting that balancing work and academics may impact performance.

## 5. Key Insights

* Students with higher engagement scores tend to achieve better grades, indicating that consistent study and attendance are critical factors.
* Attendance shows a stronger relationship with grades compared to study hours alone, suggesting consistency is more important than intensity.
* High stress and exam anxiety levels are associated with lower academic performance.
* A balanced lifestyle (adequate sleep and lower screen time) contributes positively to student outcomes.
